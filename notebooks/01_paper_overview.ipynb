## The big idea

Nested Learning reframes a model as a stack of associative-memory modules, each running its own little optimisation problem at its own update frequency. The same algebra explains:

- a single linear layer trained by SGD (Eq. 31),
- the momentum buffer in SGD-with-momentum (Eq. 33-34),
- a softmax / linear attention block (Eq. 18),
- and a self-modifying neural module that learns *how* it updates itself (Eq. 88).

Once you put many of these modules in a chain with *different update frequencies* you get the **Continuum Memory System** (CMS, §7.1) — the paper's answer to catastrophic forgetting. Compose CMS with a self-modifying "Self-Referential Titans" block and you get **HOPE** (§8, Figure 5).
